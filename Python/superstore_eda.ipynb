```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("superstore_clean.csv")
```

# Sales by Category

```python
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.show()
```

# Sales by Region

```python
df.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.show()
```

# Monthly Sales Trend

```python
monthly = df.groupby('Month')['Sales'].sum()

monthly.plot()
plt.show()
```

# Top Customers

```python
top = df.groupby('Customer_Name')['Sales'].sum()

top.sort_values(ascending=False).head(10).plot(kind='bar')
plt.show()
```

# Top Products

```python
prod = df.groupby('Product_Name')['Sales'].sum()

prod.sort_values(ascending=False).head(10).plot(kind='bar')
plt.show()
```
